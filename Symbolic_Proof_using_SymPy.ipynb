```python
import sympy as sp

# Define symbols
x, a, b, c = sp.symbols('x a b c')

# 1. Start with the standard quadratic equation: ax^2 + bx + c = 0
quadratic_eq = sp.Eq(a*x**2 + b*x + c, 0)
print("Standard form:")
sp.pprint(quadratic_eq)

# 2. Divide the equation by 'a' to isolate the x^2 coefficient
eq_div_a = sp.Eq(x**2 + (b/a)*x + (c/a), 0)

# 3. Move the constant term (c/a) to the right-hand side
eq_moved_c = sp.Eq(x**2 + (b/a)*x, -c/a)

# 4. Complete the square by adding (b / 2a)^2 to both sides
half_b_over_a_sq = (b / (2*a))**2
eq_completed_square = sp.Eq(x**2 + (b/a)*x + half_b_over_a_sq, -c/a + half_b_over_a_sq)

print("\nAfter completing the square:")
sp.pprint(eq_completed_square)

# 5. Let SymPy algebraically solve for x to extract the formal roots
solutions = sp.solve(quadratic_eq, x)

print("\nProved Mathematical Roots (The Quadratic Formula):")
for idx, sol in enumerate(solutions, 1):
    print(f"x{idx} =")
    sp.pprint(sol)
```
